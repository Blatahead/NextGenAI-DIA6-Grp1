## Part 3: Introducing the proper time and acceleration

We will find how the quantity $\frac{d^2 x_1} {d\tau^2}$ ($\tau$ is the proper time of the particle) is transformed between $R_1$ and $R_2$.

The proper time of a particle is the time that is measured by a clock that follows the movement of the particle. At each moment t_2, the derivative $\frac{dt_2} {d\tau}$ is calculated using a proper reference frame $R_1(t_2)$ (different at each time) centered at the particle immobile with respect to $R_1$ although with a speed of $v(t_2)$ with respect to $R_2$. Consequently, the time $\tau$ and $t_1$ at the origin instant have the smae derivative with respect to t_2 (which is not always the case later on since the particle might have accelerated). We have the following formulae coming from the Lorentz transformation, especially (5) that gives the time $t_{f,2}$ of a particle given its proper time $\tau_f$:

$$
\begin{align*}
\gamma &= \frac{1} {\sqrt {1 - \frac{v^2} {c^2}} } \\
\frac{dt_2} {d\tau} &= \frac{dt_2} {dt_1} = \gamma \\
T_f &= \int _0 ^{\tau_f} { \frac{1} {\sqrt {1 - \frac{v(\tau)^2} {c^2}} } d\tau} \ (5)\\
\tau_f &= \int _0 ^{t_f} {\sqrt {1 - \frac{v(t_2)^2} {c^2}} dt_2} \ (6)\\
\end{align*}
$$
