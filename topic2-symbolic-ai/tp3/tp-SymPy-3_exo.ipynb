### Results from the special relativity

The special relativity specifies how the force and the speed components of a particle p change when we observe its trajectory from the reference frame R to R',  given the following constraints:
* The reference frame R' has a constant speed $v_{R'/R}$ relatively to the reference frame R limited to its the X axis component (ie, the other components along Y and Z are zero).
* The particle p has a speed in the reference frame R $v_{p/R}$ limited to its component X (ie, the Y and Z component are zero)
* A force $F_{/R}$ observed in the R reference frame is applied to the particle p and limited to its Y component

In that condition, the speed of the particle p and the force applied to it in the R' reference frame have the following components:
$$
\begin{align*}
  \gamma &= \frac{1} {\sqrt {1 - \frac{V_{R'/R}^2} {c^2} } } \\
  F_{/R'} &= \frac{F_{/R}} {\gamma \left(1 - \frac{V_{p/R} V_{R'/R}} {c^2} \right)} \ (1) \\
  V_{p/R'} &= \frac{V_{p/R} - V_{R'/R}} {1 - \frac{V_{p/R} V_{R'/R}} {c^2} } \ (2) \\
\end{align*}
$$
